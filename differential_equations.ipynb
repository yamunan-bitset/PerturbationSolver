```python
import sympy as sp
from sympy.abc import n

poe = 3 # Power of epsilon

print('Differential Equation')
y = sp.Symbol('y')
x = sp.Symbol('x')
e = sp.Symbol('epsilon')
LHS = sp.Derivative(y, x)
RHS = y**2 + x # Input
n_expr = sp.Eq(LHS, RHS)
display(n_expr)

print("'Perturbed' Differential Equation")
pRHS = e*y**2 + x
p_expr = sp.Eq(LHS, pRHS)
display(p_expr)
```

```python
print('Set y to be the following infinite series, which is approximated as finite')

f = []
for i in range(poe+1):
    f.append(sp.Function('f_' + str(i)))
inf_series = 0
for j in range(0, poe+1):
    inf_series += f[j](x)*e**j
coeffs_inf = {}
expanded_inf_series = inf_series.doit().expand()
sp.Eq(y, expanded_inf_series)
```

```python
print("Compute perturbed equation by plugging in the infinite series for y")

lhs_eqn = LHS.subs(y, expanded_inf_series).expand().doit()
rhs_eqn = pRHS.subs(y, expanded_inf_series).expand()
sp.Eq(lhs_eqn, rhs_eqn)
```

```python
print("Equating all the like terms, and evaluating all the coefficients")

coeffs_inf = {}
eq = []
eq_s = []
C1 = sp.Symbol('C1')
for j in range(poe+1):
    eq.append(rhs_eqn.coeff(e, j))
    coeffs_inf["f" + str(j)] = sp.dsolve(sp.Eq(lhs_eqn.coeff(e, j), eq[j]), f[j](x)).simplify()
    Cn = sp.Symbol('C' + str(j))
    coeffs_inf["f" + str(j)] = coeffs_inf["f" + str(j)].subs(C1, 0) 

print(coeffs_inf)
for i in range(poe+1):
    tmp = coeffs_inf['f' + str(i)]
    for j in range(i):
        tmp = tmp.subs(f[j](x), coeffs_inf['f' + str(j)].rhs)
    coeffs_inf['f' + str(i)] = tmp.simplify()

for i in coeffs_inf:
    coeffs_inf[i] = coeffs_inf[i].rhs
    print(i + ': ' + str(coeffs_inf[i]))
```

```python
print("Finding the solution, by setting e = 1")

for j in range(poe+1):
    inf_series = inf_series.doit().subs(f[j](x), coeffs_inf['f' + str(j)])
inf_series = inf_series.subs(e, 1)
display(sp.Eq(y, sp.Poly(inf_series, x).as_expr()))
```
