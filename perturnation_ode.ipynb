```python
import sympy as sp
from sympy.abc import n, i

power_of_epsilon = 2

print("Equation:")
x = sp.Symbol('x')
y = sp.Function('y')(x)
#LHS = sp.Derivative(y, x, 2) + 4 * sp.Derivative(y, x) + 3*y + y**2
LHS = sp.diff(y, x) 
RHS = y**2 + x #0
e = sp.Symbol('epsilon')
#pLHS = sp.Derivative(y, x, 2) + 4 * sp.Derivative(y, x) + 3*y + e*y**2
pLHS = sp.diff(y) #- e*y**2 - x
pRHS = e*y**2 + x
perturbed_expr = sp.Eq(pLHS, pRHS)
display(perturbed_expr)
```

```python
print("Take taylor expansion of y with respect to epsilon")
#a = sp.IndexedBase(sp.Function("a")(x))
a = [sp.Function("a_1")(x), sp.Function("a_2")(x), sp.Function("a_3")(x)]
#inf_series = sp.Sum(a[n]*e**n, (n, 0, power_of_epsilon))
inf_series = 0
for n in range(len(a)):
    inf_series += a[n]*e**n
coeffs_inf = {}
expanded_inf_series = inf_series.doit().expand()
sp.Eq(y, expanded_inf_series)
```

```python
subs_pLHS = pLHS.subs(y, expanded_inf_series)
subs_pRHS = pRHS.subs(y, expanded_inf_series)
sp.Eq(subs_pLHS.expand(), subs_pRHS.expand())
```

```python
# Equating the terms with the same coefficient epsilon
for i in range(len(a)):
    sp.pprint(subs_pLHS.doit().coeff(sp.diff(a[i], x)))
```
